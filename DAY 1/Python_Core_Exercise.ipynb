```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise ---
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

# --- Task 1 ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# 1.1 Generate 100 simulated sensor readings
raw_sensor_data = []

base_time = datetime.datetime.now()

for i in range(100):
    timestamp = base_time + datetime.timedelta(minutes=i)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    device_id = f"DEV{i+1:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_value, max_value = READING_RANGES[device_type]
    value = round(random.uniform(min_value, max_value), 2)

    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)

print("Number of readings generated:", len(raw_sensor_data))
print("First 3 readings:")
for reading in raw_sensor_data[:3]:
    print(reading)


# 1.2 Create temperature_readings
temperature_readings = []

for timestamp, device_id, device_type, location, value in raw_sensor_data:
    if device_type == 'Temperature_Sensor':
        temperature_readings.append({
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        })

print("\nNumber of temperature readings:", len(temperature_readings))
print("First 3 temperature readings:")
for reading in temperature_readings[:3]:
    print(reading)


# 1.3 Create unique_locations
unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print("\nUnique locations:", unique_locations)


# 1.4 Create device_stats
device_stats = {}

for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

for timestamp, device_id, device_type, location, value in raw_sensor_data:
    min_value, max_value = device_stats[device_type]

    min_value = min(min_value, value)
    max_value = max(max_value, value)

    device_stats[device_type] = (min_value, max_value)

print("\nDevice statistics:")
for device_type, stats in device_stats.items():
    print(device_type, ":", stats)


# 1.5 Advanced Indexing and Slicing
print("\n--- Task 1.5: Advanced Indexing and Slicing (Lists and Strings) ---")

sample_string = "PythonProgrammingIsFun"

# 1.5.1 List slicing
# 10th to 20th inclusive means indexes 9 to 19
sliced_readings_1 = raw_sensor_data[9:20]

# Every 5th reading starting from beginning
sliced_readings_2 = raw_sensor_data[::5]

print("\nReadings from 10th to 20th:")
print(sliced_readings_1)

print("\nEvery 5th reading:")
print(sliced_readings_2)


# 1.5.2 String slicing
substring_1 = sample_string[6:17]

# Reverse of "Python"
substring_2 = sample_string[5::-1]

print("\nSubstring 1:", substring_1)
print("Substring 2:", substring_2)


# 1.5.3 Mixed slicing
fifth_reading_device_type = raw_sensor_data[4][2]
device_type_suffix = fifth_reading_device_type[-6:]

print("\nDevice type of 5th reading:", fifth_reading_device_type)
print("Last 6 characters:", device_type_suffix)


# --- Task 2 ---
print("\n--- Task 2: Function Definitions and Usage ---")

# 2.1 Function to get readings by location
def get_readings_by_location(readings_list, target_location):
    return [
        reading
        for reading in readings_list
        if reading[3] == target_location
    ]


# Test function
lab1_readings = get_readings_by_location(raw_sensor_data, 'Lab1')

print("\nNumber of readings from Lab1:", len(lab1_readings))


# 2.2 Calculate average temperature
def calculate_average_temperature(temp_data_list):
    if not temp_data_list:
        return 0.0

    total = sum(
        reading['temperature_celsius']
        for reading in temp_data_list
    )

    return total / len(temp_data_list)


average_temperature = calculate_average_temperature(temperature_readings)

print("Average temperature:", round(average_temperature, 2), "°C")


# 2.3 Generate report line
def generate_report_line(reading_dict):
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


if temperature_readings:
    print("\nSample report line:")
    print(generate_report_line(temperature_readings[0]))


# --- Task 3 ---
print("\n--- Task 3: Core Library Usage ---")

# 3.1 Create directory
reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"\nDirectory '{reports_dir}' created.")
else:
    print(f"\nDirectory '{reports_dir}' already exists.")


# 3.2 Calculate duration
if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()

    hours = int(seconds // 3600) % 24
    minutes = int((seconds % 3600) // 60)

    print(
        f"\nDuration of data collection: "
        f"{days} days, {hours} hours, {minutes} minutes."
    )
else:
    print("\nNot enough data to calculate duration.")


# 3.3 Generate 5 random integers
random_values = [random.randint(100, 200) for _ in range(5)]

print("\nRandom values:", random_values)


# --- Task 4 ---
print("\n--- Task 4: File I/O ---")

# 4.1 Write temperature readings to file
report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:
    with open(report_filename, 'w', encoding='utf-8') as f:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            f.write(line + '\n')

    print(
        f"\nTemperature report written to "
        f"'{report_filename}'."
    )

except IOError as e:
    print(
        f"\nError writing file "
        f"'{report_filename}': {e}"
    )


# 4.2 Read file contents
read_report_lines = []

try:
    with open(report_filename, 'r', encoding='utf-8') as f:
        read_report_lines = f.readlines()

    print("\nFirst 3 lines of the temperature report:")

    for line in read_report_lines[:3]:
        print(line.strip())

except IOError as e:
    print(
        f"\nError reading file "
        f"'{report_filename}': {e}"
    )


print("\n--- Exercise Complete! ---")
print("Great job completing the intermediate Python exercise.")
print("Review your code and outputs to ensure correctness and understanding of all concepts.")
print("Consider adding more complex scenarios or user inputs for further practice.")
```
